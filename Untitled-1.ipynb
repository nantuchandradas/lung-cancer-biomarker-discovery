```python

import pandas as pd 
import numpy as np 
import scipy.stats as stats
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
np.random.seed(42)

# 100 genes, 20 samples
num_genes = 100
normal_samples = [f'Normal_{i+1}' for i in range(10)]
tumor_samples = [f'Tumor_{i+1}' for i in range(10)]

# Generate baseline expression values around 10
gene_names = [f'GENE_{i+1}' for i in range(num_genes)]
data = np.random.normal(loc=10.0, scale=1.5, size=(num_genes, 20))

df = pd.DataFrame(data, index=gene_names, columns=normal_samples + tumor_samples)

# Artificially upregulate the first 5 genes in tumor samples to represent biomarkers
df.iloc[:5, 10:] += np.random.normal(loc=3.5, scale=0.5, size=(5, 10))

df.head()
```

```python
log2fc = []
pvalues = []

for gene in df.index:
    normal_vals = df.loc[gene, normal_samples]
    tumor_vals = df.loc[gene, tumor_samples]

    # Calculate Log2 Fold Change (Tumor / Normal)
    fc = tumor_vals.mean() - normal_vals.mean()
    log2fc.append(fc)

    # Welch's t-test (unequal variance)
    _, p_val = stats.ttest_ind(tumor_vals, normal_vals, equal_var=False)
    pvalues.append(p_val)

# Store results in a summary table
results = pd.DataFrame({
    'Gene': df.index,
    'log2FC': log2fc,
    'pValue': pvalues
})

# Calculate -log10(pValue) for plotting significance
results['-log10_pValue'] = -np.log10(results['pValue'])

# Print top upregulated genes
results.sort_values(by='pValue').head()
```

```python
plt.figure(figsize=(8, 6))

# Define thresholds
p_thresh = 0.05
fc_thresh = 1.0

# Categorize genes
results['Color'] = 'grey'
results.loc[(results['pValue'] < p_thresh) & (results['log2FC'] > fc_thresh), 'Color'] = 'red'    # Upregulated
results.loc[(results['pValue'] < p_thresh) & (results['log2FC'] < -fc_thresh), 'Color'] = 'blue'  # Downregulated

# Scatter plot
sns.scatterplot(data=results, x='log2FC', y='-log10_pValue', hue='Color', palette={'grey': 'grey', 'red': 'red', 'blue': 'blue'}, legend=False)

# Cutoff lines
plt.axhline(-np.log10(p_thresh), color='black', linestyle='--', linewidth=0.8)
plt.axvline(fc_thresh, color='black', linestyle='--', linewidth=0.8)
plt.axvline(-fc_thresh, color='black', linestyle='--', linewidth=0.8)

plt.title('Cancer Marker Discovery: Volcano Plot')
plt.xlabel('Log2 Fold Change (Tumor / Normal)')
plt.ylabel('-Log10 p-value')
plt.show()
```

```python
from statsmodels.stats.multitest import multipletests

# Apply Benjamini-Hochberg (FDR) adjustment
_, qvalues, _, _ = multipletests(results['pValue'], alpha=0.05, method='fdr_bh')

# Add FDR adjusted p-values (q-values) to our results table
results['qValue'] = qvalues

# Filter for genes that pass BOTH FDR < 0.05 AND Log2FC > 1.0
significant_markers = results[(results['qValue'] < 0.05) & (results['log2FC'] > 1.0)]

print(significant_markers[['Gene', 'log2FC', 'pValue', 'qValue']])
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

plt.figure(figsize=(8, 6))

# Set thresholds based on qValue (FDR)
q_thresh = 0.05
fc_thresh = 1.0

# Calculate -log10 of qValue for plotting
results['-log10_qValue'] = -np.log10(results['qValue'])

# Categorize genes using qValue
results['Color_FDR'] = 'grey'
results.loc[(results['qValue'] < q_thresh) & (results['log2FC'] > fc_thresh), 'Color_FDR'] = 'red'
results.loc[(results['qValue'] < q_thresh) & (results['log2FC'] < -fc_thresh), 'Color_FDR'] = 'blue'

# Plot
sns.scatterplot(
    data=results, 
    x='log2FC', 
    y='-log10_qValue', 
    hue='Color_FDR', 
    palette={'grey': 'grey', 'red': 'red', 'blue': 'blue'}, 
    legend=False
)

# Label the top significant genes directly on the plot
for _, row in significant_markers.iterrows():
    plt.text(row['log2FC'] + 0.1, -np.log10(row['qValue']), row['Gene'], fontsize=9, weight='bold')

plt.axhline(-np.log10(q_thresh), color='black', linestyle='--', linewidth=0.8)
plt.axvline(fc_thresh, color='black', linestyle='--', linewidth=0.8)
plt.axvline(-fc_thresh, color='black', linestyle='--', linewidth=0.8)

plt.title('Cancer Marker Discovery: FDR Corrected Volcano Plot')
plt.xlabel('Log2 Fold Change (Tumor / Normal)')
plt.ylabel('-Log10 FDR Adjusted p-value (q-value)')
plt.show()
```

```python
# Extract expression values for the significant genes across all 20 samples
top_gene_ids = significant_markers['Gene'].tolist()
heatmap_data = df.loc[top_gene_ids]

# Plot clustered heatmap
plt.figure(figsize=(10, 4))
sns.clustermap(heatmap_data, cmap='vlag', z_score=0, col_cluster=False)
plt.title('Gene Expression Heatmap (Normal vs Tumor)', pad=100)
plt.show()
```

```python
import GEOparse
import pandas as pd

# Fetch the dataset directly from NCBI GEO
gse = GEOparse.get_GEO(geo="GSE19804", destdir="./")

# Extract the gene expression table (probes x samples)
expression_df = gse.pivot_samples('VALUE')

# Display the dimensions and top rows of real data
print(f"Data shape: {expression_df.shape}")
expression_df.head()
```

```python
# Check all column names in metadata
print(gse.phenotype_data.columns.tolist())

# Inspect the first few rows to see where "normal" and "tumor" are stored
gse.phenotype_data[['title', 'source_name_ch1']].head()
```

```python
metadata = gse.phenotype_data

# Extract sample IDs based on 'source_name_ch1'
normal_samples = metadata[metadata['source_name_ch1'].str.contains('normal', case=False, na=False)].index.tolist()
tumor_samples = metadata[metadata['source_name_ch1'].str.contains('tumor', case=False, na=False)].index.tolist()

print(f"Normal Samples found: {len(normal_samples)}")
print(f"Tumor Samples found: {len(tumor_samples)}")
```

```python
import scipy.stats as stats
import numpy as np
import pandas as pd
from statsmodels.stats.multitest import multipletests

log2fc = []
pvalues = []

# Loop through all 54,675 probe IDs
for probe_id in expression_df.index:
    normal_vals = expression_df.loc[probe_id, normal_samples].dropna().astype(float)
    tumor_vals = expression_df.loc[probe_id, tumor_samples].dropna().astype(float)

    # Calculate Log2 Fold Change (Tumor - Normal)
    fc = tumor_vals.mean() - normal_vals.mean()
    log2fc.append(fc)

    # Welch's t-test
    _, p_val = stats.ttest_ind(tumor_vals, normal_vals, equal_var=False)
    pvalues.append(p_val)

# Assemble results table
results_real = pd.DataFrame({
    'Probe_ID': expression_df.index,
    'log2FC': log2fc,
    'pValue': pvalues
})

# Handle any NaN values if present
results_real['pValue'] = results_real['pValue'].fillna(1.0)

# Apply Benjamini-Hochberg (FDR) adjustment
_, qvalues, _, _ = multipletests(results_real['pValue'], alpha=0.05, method='fdr_bh')
results_real['qValue'] = qvalues

print("Analysis Complete!")
```

```python
# Extract the platform table (GPL570)
platform = gse.gpls[list(gse.gpls.keys())[0]]
gene_map = platform.table.set_index('ID')['Gene Symbol'].to_dict()

# Map Probe IDs to Gene Symbols
results_real['Gene_Symbol'] = results_real['Probe_ID'].map(gene_map)

# Filter for top upregulated lung cancer biomarkers (qValue < 0.01 & Log2FC > 1.5)
top_cancer_markers = results_real[
    (results_real['qValue'] < 0.01) & 
    (results_real['log2FC'] > 1.5) & 
    (results_real['Gene_Symbol'].notna()) &
    (results_real['Gene_Symbol'] != '')
].sort_values(by='qValue')

# Display top 10 real cancer marker candidates
top_cancer_markers[['Probe_ID', 'Gene_Symbol', 'log2FC', 'pValue', 'qValue']].head(10)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Prepare log-transformed q-values
results_real['-log10_qValue'] = -np.log10(results_real['qValue'])

plt.figure(figsize=(10, 7))

# Thresholds: Log2FC > 1.5, qValue < 0.01
q_cutoff = 0.01
fc_cutoff = 1.5

# Categorize points
results_real['Color'] = 'grey'
results_real.loc[(results_real['qValue'] < q_cutoff) & (results_real['log2FC'] > fc_cutoff), 'Color'] = 'red'
results_real.loc[(results_real['qValue'] < q_cutoff) & (results_real['log2FC'] < -fc_cutoff), 'Color'] = 'blue'

# Scatter plot
sns.scatterplot(
    data=results_real, 
    x='log2FC', 
    y='-log10_qValue', 
    hue='Color', 
    palette={'grey': '#cccccc', 'red': '#e41a1c', 'blue': '#377eb8'}, 
    alpha=0.6,
    s=15,
    legend=False
)

# Label top 5 genes directly on the plot
top_label_genes = top_cancer_markers.head(5)
for _, row in top_label_genes.iterrows():
    y_pos = -np.log10(row['qValue'])  # Calculate directly from qValue
    plt.text(
        row['log2FC'] + 0.1, 
        y_pos, 
        str(row['Gene_Symbol']), 
        fontsize=9, 
        weight='bold'
    )

# Threshold lines
plt.axhline(-np.log10(q_cutoff), color='black', linestyle='--', linewidth=0.8)
plt.axvline(fc_cutoff, color='black', linestyle='--', linewidth=0.8)
plt.axvline(-fc_cutoff, color='black', linestyle='--', linewidth=0.8)

plt.title('GSE19804: Differential Gene Expression (Lung Tumor vs Normal)', fontsize=12)
plt.xlabel('Log2 Fold Change (Tumor / Normal)', fontsize=10)
plt.ylabel('-Log10 FDR Adjusted p-value (q-value)', fontsize=10)
plt.show()
```
